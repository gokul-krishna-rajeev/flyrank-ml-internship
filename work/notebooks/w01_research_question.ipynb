## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Verify starter dataset exists and can be read
import os
import pandas as pd

# Use raw.githubusercontent.com for direct raw CSV access
DATA_PATH = "https://raw.githubusercontent.com/gokul-krishna-rajeev/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

# Verify connection
df_check = pd.read_csv(DATA_PATH, nrows=5)
print(f"Data connection verified. Columns available: {len(df_check.columns)}")

# Check grain uniqueness and target proxy distribution
df_full = pd.read_csv(DATA_PATH)

is_unique_grain = df_full['content_id'].nunique() == len(df_full)
print(f"Grain Check - Is content_id 100% unique per row? {is_unique_grain}")
print(f"Target Proxy ('trend_direction' == 'down') distribution:")
print(df_full['trend_direction'].value_counts(normalize=True))

# Generate the 3 real supporting numbers from the starter dataset
df = pd.read_csv(DATA_PATH)

# Number 1: Dataset size and client count
total_rows = len(df)
total_clients = df['client_id'].nunique()

# Number 2: Zero position missingness count
zero_pos_count = (df['avg_position'] == 0).sum()

# Number 3: Baseline rule precision estimation vs dataset totals
stale_visible = ((df['days_since_last_update'] >= 180) & (df['impressions_90d'] >= 500)).sum()

print(f"--- STARTER DATA EVIDENCE ---")
print(f"1. Total Content Rows: {total_rows:,} across {total_clients} clients")
print(f"2. Rows with avg_position == 0 (Missing Data): {zero_pos_count:,} ({zero_pos_count/total_rows:.1%})")
print(f"3. High-demand stale pages (Impressions >= 500 & Age >= 180d): {stale_visible:,}")
```
